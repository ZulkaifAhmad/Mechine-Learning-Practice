```python
from scipy.stats import pearsonr
# pearsonr is a function from the SciPy (Scientific Python) library.
# It is used to measure the linear relationship between two numeric variables.
# Pearson only detects linear relation
# Find which features affect intelligence_rating
# intelligence_rating is the output (target) feature.
targeted_column = df['intelligence_rating']

extract_colums = [ 'alignment', 'gender', 'first_appearance_year', 'powers_count',
       'has_powers', 'strength_rating',
       'movie_appearances', 'species_asgardian',
       'species_cyborg', 'species_destroyer', 'species_enhanced human',
       'species_flora colossus', 'species_frost giant',
       'species_genetically engineered raccoon', 'species_human',
       'species_human-celestial hybrid', 'species_human-kree hybrid',
       'species_mutated human', 'species_synthetic', 'species_titan',
       'species_zehoberei', 'affiliation_Asgard', 'affiliation_Avengers',
       'affiliation_Black Order', 'affiliation_Guardians',
       'affiliation_SHIELD', 'affiliation_Wakanda', 'affiliation_none','box_office_debut_million']

coorelation = {
    col: pearsonr(df[col], targeted_column)[0]  # [0] means → take only correlation value, ignore p-value
    for col in extract_colums
    if df[col].dtype != 'object'   # ignore string columns
}

coorelation_df = pd.DataFrame(list(coorelation.items()), columns=['features', 'correlated'])

coorelation_df = coorelation_df.sort_values(by='correlated', ascending=False)

coorelation_df
```

```python
from scipy.stats import chi2_contingency
# chi2_contingency is a function from SciPy (Scientific Python) used to check:
# Is there a relationship between two categorical (group-based) variables?

alpha = 0.1 # This is your threshold
# ✔ If p_value < 0.1 → feature is important


binned_target = pd.qcut(df['intelligence_rating'] , q=4 , labels=False)
# qcut : Converts numeric data into categories (groups)

chi2_result = {}

for col in extract_colums:
    contingency = pd.crosstab(df[col] , binned_target) 
    # It creates a frequency table
    
#       | gender | Revenue Category |
#       | ------ | ---------------- |
#       | male   | 0                |
#       | female | 1                |

#       after crosstab it becomes : 
#       | gender | 0  | 1 |
#       | ------ | -- | - |
#       | male   | 10 | 5 |
#       | female | 7  | 8 |


    chi2_state , p_value , dof , excepted = chi2_contingency(contingency);
    
    # chi2_state : Strength of difference
    # p_value : Significance (important)
    # dof : Degrees of freedom
    # expected : Expected values
    
    if p_value < alpha:
        
        decision = "Keep Feature"
    else:
        decision = "Drop Feature"
    chi2_result[col] = {
        "Chi_2_Statistics" : chi2_state ,
        "p_value" : p_value ,
        "Decision" : decision ,
        'dof' : dof ,
        'excepted' : excepted
    }
#    dof :  
# It just tells you how many independent values are free to change in your data.
# It’s a property of your dataset, not a score.

chi2_df = pd.DataFrame(chi2_result).T
chi2_df = chi2_df.sort_values(by="p_value")
chi2_df
```
