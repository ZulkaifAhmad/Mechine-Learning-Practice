```python
y = df['price']
X = df.drop(columns=['price'])
X.head()
```

```python
X_label_encoding = X.copy()
X_oneHot_encoding = X.copy()
```

```python
X_label_encoding.columns
```

```python
# label encoding

from sklearn.preprocessing import LabelEncoder
features = ['model', 'year', 'transmission' ,  'mileage', 'fuelType', 'tax', 'mpg','engineSize']
for i in features:
    label_encoder = LabelEncoder()
    X_label_encoding[i] = label_encoder.fit_transform(X_label_encoding[i])
    
```

```python
X_label_encoding.head()
```

```python
# one hot encoding 
X_oneHot_encoding = pd.get_dummies(X_oneHot_encoding, columns=['model', 'transmission', 'fuelType'], drop_first=True)
X_oneHot_encoding.head()
```

```python
X_oneHot_encoding = X_oneHot_encoding.astype(int)
```

```python
X_oneHot_encoding.head()
```

```python
X_oneHot_encoding.info()
```

```python
X_label_encoding.columns
```

```python
# Scaling : 

from sklearn.preprocessing import StandardScaler

scaling_features = ['year' ,'mileage' ,	'tax' ,	'mpg']
scaler_oneHot = StandardScaler()

X_oneHot_encoding[scaling_features] = scaler_oneHot.fit_transform(X_oneHot_encoding[scaling_features])

for i in features:
    scaler_label = StandardScaler()
    X_label_encoding[[i]] = scaler_label.fit_transform(X_label_encoding[[i]])
```

```python
X_label_encoding.head()
```

```python
X_oneHot_encoding.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_oneHot_encoding, y, test_size=0.20, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train , y_train)
```

```python
y_predict = model.predict(X_test)
r2 = r2_score(y_test , y_predict)
r2
```

```python
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1 - ((1- r2) * (n-1) / (n - p - 1))
adjusted_r2
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_label_encoding, y, test_size=0.20, random_state=42)
```

```python
label_model = LinearRegression()
label_model.fit(X_train , y_train)
```

```python
label_y_predict = label_model.predict(X_test)
r2_label = r2_score(y_test , label_y_predict)
r2_label
```

```python
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_label_r2 = 1 - ((1- r2_label) * (n-1) / (n - p - 1))
adjusted_label_r2
```

```python
print(f"r2_score for label_encoded data : {r2_label}")
print(f"Adjusted r2_score for lable_encoded data : {adjusted_label_r2}")
print(f"r2_score for one_hot_encoded data : {r2}")
print(f"Adjsted r2_score for one_hot_encoded data : {adjusted_r2}")
```
